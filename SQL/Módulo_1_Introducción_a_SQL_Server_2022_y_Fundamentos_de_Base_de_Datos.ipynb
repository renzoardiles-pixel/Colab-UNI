# **<font color="#FFFFFF">¿Qué es SQL Server 2025?</font>**

</td>
</tr>
</table>

SQL Server 2025 es la nueva versión (2025) del sistema de bases de datos relacional de Microsoft, diseñada para empresas que necesitan gestionar, almacenar y consultar datos de forma segura, eficiente y escalable.

Más que una simple actualización, esta versión representa un salto importante hacia la integración de la inteligencia artificial (IA) directamente dentro del motor de la base de datos.

SQL Server es un sistema de gestión de bases de datos (DBMS) que permite almacenar, organizar y consultar datos usando el lenguaje T-SQL (Transact-SQL).

# **<font color="#FFFFFF">¿Qué trae de nuevo SQL Server 2025?</font>**

</td>
</tr>
</table>

>* **1. IA integrada nativamente**: SQL Server 2025 incorpora capacidades de IA directamente en la base de datos, permitiendo:
    * **Búsqueda semántica y vectores**: puedes usar búsquedas inteligentes basadas en similitud y significado, no solo texto exacto.
    * **Soporte para modelos de IA** mediante REST APIs, lo que facilita utilizar modelos de Azure OpenAI u otros sin cambiar tu código SQL.
    * Esto abre grandes posibilidades para aplicaciones que necesitan análisis semántico, chat inteligente con datos o RAG (Recuperación Aumentada por IA) directamente desde la base de datos.

>* **2. Mejoras para desarrolladores**: SQL Server 2025 incluye mejoras que agilizan el desarrollo:
    * **Soporte nativo de JSON** para trabajar con datos semiestructurados.
    * **Expresión regular (RegEx)** y funciones para manipular texto.
    * **Streaming de eventos de cambio**, para aplicaciones en tiempo real.
    * **Soporte para REST APIs y herramientas modernas** que facilitan conectar otros servicios o lógica externa.

>* **3. Rendimiento, escalabilidad y seguridad**: Aunque SQL Server siempre ha sido fuerte en estas áreas, la versión 2025 mejora:
    * **Mejoras en el procesamiento de consultas** para más velocidad.
    * **Autenticación moderna con identidades administradas** y mejor seguridad en entornos híbridos.
    * **Escalabilidad ampliada**, por ejemplo con Standard Edition soportando hasta 32 núcleos y 256 GB de RAM.

>* **4. Integración con la nube**: SQL Server 2025 fue diseñado para funcionar sin problemas tanto:
    * **On-premises** (en tus propios servidores)
    * **En la nube**, especialmente con Microsoft Azure, Azure Arc y mirroring con Azure Fabric.
    * Esto facilita gestionar tus bases de datos de forma centralizada sin importar dónde estén ejecutándose.

>* **5. Ediciones disponibles**: Como versiones anteriores, SQL Server 2025 viene en varias ediciones pensadas para diferentes necesidades:
    * **Express** - Gratis y ligera (hasta ~50 GB).
    * **Standard** - Para aplicaciones de negocios medianos.
    * **Enterprise** - Para cargas de trabajo de misión crítica.
    * **Developer** - Gratis para desarrollar y probar.

## **<font color="#FFFFFF">Motor de Base de Datos</font>**

</td>
</tr>
</table>

El Motor de Base de Datos es el núcleo de SQL Server y se encarga de las operaciones fundamentales de almacenamiento, recuperación y manipulación de datos.

Sus funciones principales incluyen:

>* **Gestión de Datos**: Permite crear, leer, actualizar y eliminar datos en bases de datos relacionales.
>* **Transacciones**: Asegura que las operaciones sobre la base de datos sean ejecutadas de manera coherente y segura a través de transacciones ACID (Atomicidad, Consistencia, Aislamiento, Durabilidad).
>* **Seguridad**: Implementa controles de acceso y autenticación para proteger los datos.
>* **Optimización de Consultas**: Mejora el rendimiento a través de un optimizador de consultas que elige el mejor plan de ejecución.
>* **Soporte para Indices**: Utiliza índices para acelerar la búsqueda y recuperación de datos.

## **<font color="#FFFFFF">SQL Server Management Studio (SSMS)</font>**

</td>
</tr>
</table>

**SQL Server Management Studio** es una herramienta gráfica de administración y desarrollo para SQL Server. Proporciona una interfaz amigable para interactuar con el Motor de Base de Datos.

Algunas de sus características son:

>* **Explorador de Objetos**: Permite explorar y administrar todos los objetos en la base de datos (tablas, vistas, procedimientos almacenados, etc.).
>* **Editor de consultas**: Herramienta para escribir y ejecutar sentencias T-SQL. Ofrece funcionalidades como resaltado de sintaxis y completado automático.
>* **Gestión de Seguridad**: Facilita la gestión de permisos y roles de usuarios.
>* **Herramientas de diagnóstico**: Incluye herramientas para monitorear el rendimiento y solucionar problemas en la base de datos.
>* **Opciones de Backup y Restore**: Permite realizar copias de seguridad y restaurar bases de datos fácilmente.

### **<font color="#FFFFFF">1. SQL Server Express</font>**

</td>
</tr>
</table>

* **Descripción**: SQL Server Express es una base de datos básica gratuita que es adecuada para construir aplicaciones de escritorio y pequeñas aplicaciones de servidor basadas en datos, lo que la convierte en una opción popular para proveedores de software independientes, desarrolladores y aficionados que construyen aplicaciones cliente. Puede actualizarse sin problemas a otras versiones de gama alta de SQL Server si se necesitan características avanzadas de base de datos. SQL Server Express LocalDB es una versión liviana de la edición Express que tiene todas sus características de programación, se ejecuta en modo usuario, tiene una instalación rápida y sin configuración, y requiere solo una breve lista de requisitos previos.
* **Uso Ideal**: Base de datos gratuita y de nivel inicial para pequeñas aplicaciones de IA, web y móviles.
* **Características**:
    * Tamaño máximo de bases de datos relacionales: Ahora hasta 50 GB (NUEVA)
    * Capacidad de CPU: Compatible con hasta 4 núcleos
    * Memoria: Hasta 1410 MB de memoria
    * Búsqueda vectorial y semántica: Integración de búsqueda nativa (NUEVA)
    * Soporte nativo para JSON: Ahora se incluye un tipo de dato JSON (NUEVA)
    * Invocación de puntos finales REST externos: Nueva funcionalidad para interactuar con servicios externos (NUEVA)
    * Transmisión de eventos de cambios: Capacidad para realizar streaming de eventos (NUEVA)
    * Autenticación con Microsoft Entra ID: Nuevo método de autenticación segura (NUEVA)
    * Espejado en Fabric: Introducción de nuevas opciones de espejado (NUEVA)

#### **<font color="#FFFFFF">Habilitar Puertos SQL Server 2025 en Producción</font>**

</td>
</tr>
</table>



```
@echo on
cls
@echo off
echo =========================================
echo Configurando Firewall para SQL Server 2025
echo =========================================

:: SQL Server – Puerto fijo TCP 1433
netsh advfirewall firewall add rule name="SQL Server TCP 1433" ^
    dir=in action=allow protocol=TCP localport=1433

:: Dedicated Admin Connection (DAC)
netsh advfirewall firewall add rule name="SQL DAC TCP 1434" ^
    dir=in action=allow protocol=TCP localport=1434

:: SQL Service Broker (opcional)
netsh advfirewall firewall add rule name="SQL Service Broker TCP 4022" ^
    dir=in action=allow protocol=TCP localport=4022

:: SQL Browser (solo si usas instancias con puertos dinámicos)
netsh advfirewall firewall add rule name="SQL Browser UDP 1434" ^
    dir=in action=allow protocol=UDP localport=1434

:: Analysis Services
netsh advfirewall firewall add rule name="SSAS TCP 2383" ^
    dir=in action=allow protocol=TCP localport=2383

netsh advfirewall firewall add rule name="SSAS Browser TCP 2382" ^
    dir=in action=allow protocol=TCP localport=2382

:: HTTP / HTTPS (solo si aplica)
netsh advfirewall firewall add rule name="HTTP TCP 80" ^
    dir=in action=allow protocol=TCP localport=80

netsh advfirewall firewall add rule name="HTTPS TCP 443" ^
    dir=in action=allow protocol=TCP localport=443

echo Firewall configurado correctamente.

pause

```
