## **<font color="#FFFFFF">IIF(Condición, Verdadero, Falso)</font>**

</td>
</tr>
</table>

La función IIF() en SQL Server es una función condicional que permite evaluar una expresión y devolver un valor basado en el resultado de esa evaluación.

Es similar a la función CASE, pero más compacta y fácil de usar en situaciones simples.



```
SELECT
      IIF([Title] IS NULL, '' , [Title] + ' ') + [FirstName] +
      IIF([MiddleName] IS NULL, '' , ' ' + [MiddleName]) + ' ' + [LastName] +
      IIF([Suffix] IS NULL, '' , ' ' + [Suffix]) AS FullName
FROM [AdventureWorks2025].[Person].[Person]
WHERE [Suffix] IS NOT NULL
```

## **<font color="#FFFFFF">CASE</font>**

</td>
</tr>
</table>

La expresión CASE en SQL Server es una forma de manejar la lógica condicional dentro de las consultas.

Permite evaluar condiciones y devolver un valor basado en el resultado de esas evaluaciones.

Es útil cuando necesitas ejecutar más de una comparación y es más flexible que la función IIF().

USE master
GO
SELECT DISTINCT
	xtype AS Tipo,
	CASE xtype
		WHEN 'AF' THEN 'Función de agregado (CLR)'

		WHEN 'F' THEN 'Restricción FOREIGN KEY'

		WHEN 'P' THEN 'Procedimiento almacenado'
		WHEN 'PC' THEN 'Procedimiento almacenado del ensamblado (CLR)'
		WHEN 'PK' THEN 'Restricción PRIMARY KEY (de tipo K)'
		WHEN 'RF' THEN 'Procedimiento almacenado de filtro de replicación'

## **<font color="#FFFFFF">OPERADORES DE RELACIÓN O COMPARACIÓN</font>**

| Operador | Descripción |
|---|---|
| = | Igual a |
| > | Mayor que |
| < | Menor que |
| >= | Mayor o igual que |
| <= | Menor o igual que |
| <> | Diferente de |
| != | Diferente de |
| !> | No mayor que |
| !< | No menor que |
| ALL | Compara un valor escalar con un conjunto de valores de una sola columna. |
| ANY / SOME | Compara un valor escalar con cualquier valor de un conjunto de valores de una sola columna. |
| BETWEEN | Indica si un valor está dentro de un rango especificado. |
| EXISTS | Se utiliza para comprobar la existencia de cualquier fila en una subconsulta. |
| IN | Compara un valor escalar con una lista de valores o el resultado de una subconsulta. |
| LIKE | Compara un valor de cadena con un patrón especificado. |
| NOT | Se utiliza para negar el resultado de una condición. |
| OR | Combina dos o más condiciones y devuelve verdadero si al menos una de las condiciones es verdadera. |
| IS NULL | Se utiliza para comprobar si un valor es NULL. |
| IS NOT NULL | Se utiliza para comprobar si un valor no es NULL. |



```
-- CALCULAR LA EDAD DE LOS BENEFICIARIOS AL 30 DE SETIEMBRE DE 2004
-- Y LISTARLOS CON SUS DATOS PERSONALES
USE pvl
GO
SELECT
    IIF(TRIM(pat_ben_vc) = '', '', pat_ben_vc + ' ') +
    mat_ben_vc + ', ' +
    nom_ben_vc AS BENEFICIARIO,
    dni_ben_ch AS DNI,
    DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004  AS EDAD1,
    CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004)  AS EDAD2,
    fec_emb_ben_da AS FECHA_EMB,
    fec_lac_ben_da AS FECHA_LAC,
    IIF(cod_sex_bi = 0, 'FEMENINO', 'MASCULINO') AS SEXO,
    CASE
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) = 0 THEN 'MENORES DE 1 AÑO'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) BETWEEN 1 AND 6  THEN 'NIÑOS DE 1 A 6 AÑOS'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) BETWEEN 7 AND 13  THEN '7 A 13 AÑOS'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) BETWEEN 14 AND 17 THEN 'ADOLESCENTE'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) BETWEEN 18 AND 29 THEN 'JOVEN'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004) BETWEEN 30 AND 59 THEN 'ADULTO'
        WHEN CONVERT(INT, DATEDIFF(DAY, fec_nac_ben_da, '2004-09-30') / 365.256363004)>= 60 THEN 'ADULTO MAYOR'
        ELSE 'ERROR'
    END AS [CICLO DE VIDA]
FROM Beneficiario
GO
```
