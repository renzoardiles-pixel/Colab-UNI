# **<font color="#FFFFFF">¿QUÉ ES UNA BASE DE DATOS?</font>**

</td>
</tr>
</table>

Una base de datos es un conjunto organizado de datos almacenados electrónicamente y accesibles para su uso. Una base de datos permite la recuperación, modificación y administración de los datos de manera eficiente y segura.

Hay muchos tipos diferentes de bases de datos, incluyendo:
>* **Bases de datos relacionales**: almacenan los datos en tablas relacionadas y utilizan un lenguaje de consulta estructurado (SQL) para acceder a los datos.
>* **Bases de datos NoSQL**: permiten almacenar y acceder a los datos sin un esquema fijo y a menudo se utilizan para aplicaciones de big data y aplicaciones en tiempo real.
>* **Bases de datos de documentos**: almacenan los datos en forma de documentos y se utilizan para aplicaciones que requieren flexibilidad en la estructura de los datos.
>* **Bases de datos de grafos**: almacenan y permiten acceder a los datos en forma de nodos y relaciones y se utilizan para aplicaciones que requieren modelar relaciones complejas.

En cualquier caso, las bases de datos se utilizan en una amplia variedad de aplicaciones, desde sistemas de información empresarial hasta aplicaciones web y móviles. Su uso permite una mejor organización, accesibilidad y seguridad de los datos, lo que facilita la toma de decisiones y mejora la eficiencia en la gestión de la información.

## **<font color="#FFFFFF">Base de Datos Master</font>**

</td>
</tr>
</table>

La base de datos master almacena toda la información de configuración y metadatos que son necesarios para que el servidor SQL funcione correctamente.

Esto incluye las configuraciones globales y los detalles sobre todas las otras bases de datos en el servidor.

>* **Contenido de la Base de Datos Master**
    * **Información de Sistemas**: Almacena datos sobre el estado del servidor, configuraciones, y recursos, como:
        * Configuraciones de servidor (ej. opciones de recuperación, límites de memoria).
        * Información sobre usuarios y roles de seguridad.
        * Detalles sobre las bases de datos en el servidor y sus propiedades (nombre, tamaño, estado, etc.).
    * **Registros de Seguridad**: Información sobre las cuentas de inicio de sesión, roles de servidor y permisos otorgados a cada usuario.
    * **Configuraciones del Servidor**: Contiene información sobre los parámetros de configuración del servidor, como la configuración de red y las opciones de ejecución.



>* **Funcionalidades Clave**
    * **Gestión de Bases de Datos**: Permite la creación, eliminación y configuración de otras bases de datos del servidor. Cuando creas o eliminas una base de datos, esta información se refleja en la base de datos master.
    * **Control de Seguridad**: Maneja los inicios de sesión y permisos en el servidor. Sin los registros en la base de datos master, no se podrían autenticar los usuarios ni gestionar los accesos.
    * **Recovery y Backup**: Es crucial para la recuperación del sistema. Si la base de datos master se corrompe, esto puede poner en riesgo la capacidad del servidor para funcionar, por lo que se recomienda realizar copias de seguridad regularmente.

## **<font color="#FFFFFF">Base de Datos Model</font>**

</td>
</tr>
</table>

La base de datos **model** en SQL Server es una base de datos especial que sirve como plantilla o modelo para crear nuevas bases de datos.

>* **1. Propósito Principal**
    * La base de datos model actúa como un prototipo. Cuando creas una nueva base de datos, SQL Server copia la estructura y configuración de la base de datos model para inicializar la nueva base de datos.

>* **2. Contenido de la Base de Datos Model**
    * **Estructura Base**: Puedes predefinir tablas, vistas, procedimientos almacenados, y otros objetos que deseas que todas las nuevas bases de datos hereden.
    * **Configuraciones Predeterminadas**: Cualquier configuración específica, como niveles de recuperación o parámetros de compatibilidad, se puede establecer en la base de datos model.

>* **3. Funcionalidades Clave**
    * *Plantilla para Nuevas Bases de Datos**: Al crear una nueva base de datos, SQL Server usa la base de datos model para establecer la estructura inicial.

>* **msdb**: Almacena toda la data que utiliza el SQL Server Agent para programar alertas y trabajos, y para registrar operadores.
>* **tempdb**: Para almacenamiento de tablas temporales.

```
USE master
GO
IF EXISTS
(
	SELECT name
	FROM sysdatabases
	WHERE name = 'empresa'
)
DROP DATABASE [empresa]
GO
CREATE DATABASE [empresa]
ON  PRIMARY
(
	NAME = N'empresa_dat',
	FILENAME = N'H:\db\empresa_dat.mdf' ,
	SIZE = 30720KB ,
	MAXSIZE = 51200KB ,
	FILEGROWTH = 10240KB
)
LOG ON
(
	NAME = N'empresa_log',
	FILENAME = N'J:\db\empresa_log.ldf' ,
	SIZE = 10240KB ,
	FILEGROWTH = 10%
)
```





```
USE master
GO
IF EXISTS
(
	SELECT name
	FROM sysdatabases
	WHERE name = 'BCP'
)
DROP DATABASE BCP
GO
CREATE DATABASE BCP
ON PRIMARY
(
	name = bcp_dat,
	filename = 'C:\db\bcp_dat.mdf',
	size = 30MB,
	maxsize=50MB,
	filegrowth = 10MB
)
LOG ON
(
	name = bcp_log,
	filename = 'C:\db\bcp_log.ldf',
	size = 10MB,
	filegrowth = 10%,
	maxsize=UNLIMITED
)
```

# 📘 Vistas de Compatibilidad en SQL Server

Las **vistas de compatibilidad** son objetos heredados de versiones antiguas (SQL Server 2000 y anteriores).
Microsoft las mantiene **solo por retrocompatibilidad** y están **obsoletas**.

> ⚠️ **No se recomienda su uso en desarrollos nuevos**
> ✅ Usar siempre las vistas modernas del esquema `sys`

---

## 📌 Características

* No pertenecen a un esquema (`sysobjects`, no `sys.objects`)
* Simulan el comportamiento de tablas del sistema antiguas
* Pueden eliminarse en versiones futuras

---

## 📂 Listado de vistas de compatibilidad

---

## `sysdatabases`

**Descripción:**
Información de todas las bases de datos del servidor.

**Vista moderna equivalente:**
`sys.databases`

```sql
SELECT *
FROM sysdatabases
```

---

## `sysobjects`

**Descripción:**
Contiene todos los objetos de la base de datos (tablas, vistas, procedimientos, funciones, triggers).

**Vista moderna equivalente:**
`sys.objects`

```sql
-- Tablas de usuario
SELECT *
FROM sysobjects
WHERE xtype = 'U' AND uid = 1
```

**Valores comunes de `xtype`:**

* `U` → Tabla de usuario
* `V` → Vista
* `P` → Procedimiento almacenado
* `FN` → Función escalar
* `TR` → Trigger

---

## `syscolumns`

**Descripción:**
Información de columnas de tablas y vistas.

**Vista moderna equivalente:**
`sys.columns`

```sql
SELECT *
FROM syscolumns
WHERE id = OBJECT_ID('MiTabla')
```

---

## `systypes`

**Descripción:**
Tipos de datos del sistema y definidos por el usuario.

**Vista moderna equivalente:**
`sys.types`

```sql
SELECT *
FROM systypes
```

---

## `sysindexes`

**Descripción:**
Índices y estadísticas de tablas y vistas.

**Vista moderna equivalente:**
`sys.indexes`

```sql
SELECT *
FROM sysindexes
WHERE id = OBJECT_ID('MiTabla')
```

---

## `sysusers`

**Descripción:**
Usuarios definidos en la base de datos.

**Vista moderna equivalente:**
`sys.database_principals`

```sql
SELECT *
FROM sysusers
```

---

## `syslogins`

**Descripción:**
Logins del servidor SQL Server.

**Vista moderna equivalente:**
`sys.server_principals`

```sql
SELECT *
FROM syslogins
```

---

## `sysfiles`

**Descripción:**
Archivos físicos asociados a la base de datos.

**Vista moderna equivalente:**
`sys.database_files`

```sql
SELECT *
FROM sysfiles
```

---

## `sysfilegroups`

**Descripción:**
Filegroups definidos en la base de datos.

**Vista moderna equivalente:**
`sys.filegroups`

```sql
SELECT *
FROM sysfilegroups
```

---

## `sysprocesses`

**Descripción:**
Procesos y conexiones activas al servidor.

**Vista moderna equivalente:**
`sys.dm_exec_sessions`

```sql
SELECT *
FROM sysprocesses
WHERE spid > 50
```

Texto fuente de procedimientos, vistas y funciones.

Restricciones `CHECK` y `DEFAULT`.

## `sysreferences`

**Descripción:**
Relaciones de claves foráneas.

**Vista moderna equivalente:**

# **<font color="#FFFFFF">Esquemas</font>**

</td>
</tr>
</table>

Un esquema es una colección de objetos que pertenecen a una misma entidad y que se organizan de manera lógica.

Proporciona una forma de agrupar y categorizar los objetos dentro de una base de datos, como tablas, vistas, procedimientos almacenados, funciones y más.

<table width='99%'>
<tr>
<td bgcolor='#5b2c6f'>

## **<font color="#FFFFFF">Propósito de los Esquemas</font>**

</td>
</tr>
</table>

>* **Organización**: Los esquemas permiten organizar objetos de manera que se puedan gestionar más fácilmente, especialmente en bases de datos grandes con muchos componentes.
>* **Seguridad**: Facilitan la implementación de seguridad, ya que los permisos pueden ser asignados a un esquema completo, controlando el acceso a todos sus objetos simultáneamente.
>* **Colaboración**: Diferentes esquemas pueden ser utilizados para diferentes aplicaciones o módulos dentro de una misma base de datos, permitiendo un desarrollo más ordenado y colaborativo.

Cuando no se especifica un esquema al crear un objeto, se asigna automáticamente al esquema dbo (database owner). Este es el esquema por defecto en SQL Server.

## **<font color="#FFFFFF">¿Qué es una Tabla?</font>**

</td>
</tr>
</table>

Una tabla es un objeto de la base de datos que se utiliza como almacenamiento de los datos.

Una tabla tiene un formato fila-columna similar a las tablas o listas de Microsoft Excel. Cada fila es un registro único en la tabla (no existe una fila igual a otra en la misma tabla), y cada columna almacena los valores de propiedades específicas del registro. Por ejemplo, si tenemos una tabla EMPLEADO, cada fila almacena los datos de un único empleado, y en cada columna podemos almacenar su apellido paterno, su apellido materno, su nombre, su fecha de nacimiento, su estado civil y número de hijos, por lo que antes de crear una tabla se debe definir el tipo de dato que almacenará cada columna. Un tipo de dato define el tipo de información (cadena, número o fecha) que la columna almacena.

Los tipos de datos creados por el usuario permiten definir tipos de datos personalizados que pueden ser utilizados en tablas, columnas, variables y parámetros los cuales derivan de los tipos de datos de SQL Server 2025.

Esto proporciona un alto grado de flexibilidad y ayuda a asegurar la integridad de los datos.



```
USE empresa
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_codigo'
)
BEGIN
	DROP TYPE td_codigo
END
GO
CREATE TYPE td_codigo FROM INT
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_apellido'
)
BEGIN
	DROP TYPE td_apellido
END
GO
CREATE TYPE td_apellido FROM VARCHAR(40)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_nombre'
)
BEGIN
	DROP TYPE td_nombre
END
GO
CREATE TYPE td_nombre FROM VARCHAR(35)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_dni'
)
BEGIN
	DROP TYPE td_dni
END
GO
CREATE TYPE td_dni FROM CHAR(8)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_fecha_nacimiento'
)
BEGIN
	DROP TYPE td_fecha_nacimiento
END
GO
CREATE TYPE td_fecha_nacimiento FROM DATE
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_razon_social'
)
BEGIN
	DROP TYPE td_razon_social
END
GO
CREATE TYPE td_razon_social FROM VARCHAR(150)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_ruc'
)
BEGIN
	DROP TYPE td_ruc
END
GO
CREATE TYPE td_ruc FROM CHAR(11)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_direccion'
)
BEGIN
	DROP TYPE td_direccion
END
GO
CREATE TYPE td_direccion FROM VARCHAR(200)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_telefono'
)
BEGIN
	DROP TYPE td_telefono
END
GO
CREATE TYPE td_telefono FROM CHAR(9)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_email'
)
BEGIN
	DROP TYPE td_email
END
GO
CREATE TYPE td_email FROM VARCHAR(100)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_fecha_contratacion'
)
BEGIN
	DROP TYPE td_fecha_contratacion
END
GO
CREATE TYPE td_fecha_contratacion FROM DATE
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_salario'
)
BEGIN
	DROP TYPE td_salario
END
GO
CREATE TYPE td_salario FROM DECIMAL(10,2)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_cargo'
)
BEGIN
	DROP TYPE td_cargo
END
GO
CREATE TYPE td_cargo FROM VARCHAR(50)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_producto'
)
BEGIN
	DROP TYPE td_producto
END
GO
CREATE TYPE td_producto FROM VARCHAR(100)
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_sexo'
)
BEGIN
	DROP TYPE td_sexo
END
GO
CREATE TYPE td_sexo FROM BIT
GO
IF EXISTS
(
	SELECT name
	FROM sys.types
	WHERE name = 'td_estado'
)
BEGIN
	DROP TYPE td_estado
END
GO
CREATE TYPE td_estado FROM INT
GO
SELECT
	t2.name + ':'+ t1.name+
	CASE t2.precision + t2.scale
		WHEN 0 THEN '(' + LTRIM(t2.max_length) + ')'
		ELSE
			CASE
				WHEN t2.scale > 0 THEN '(' + LTRIM(t2.precision) + ', ' + LTRIM(t2.scale) + ')'
				ELSE ''
			END
	END AS tipo_definido,
	t2.max_length,
	t2.precision,
	t2.scale
FROM sys.types t1
INNER JOIN sys.types t2
ON t1.user_type_id = t2.system_type_id
WHERE t2.name LIKE 'td%'
ORDER BY t2.name
GO
```
