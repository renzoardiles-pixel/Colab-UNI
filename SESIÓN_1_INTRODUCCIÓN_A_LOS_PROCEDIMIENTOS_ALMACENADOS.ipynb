# **<font color="#FFFFFF">¿Qué son los procedimientos almacenados?</font>**

</td>
</tr>
</table>

Los procedimientos almacenados son conjuntos de instrucciones T-SQL que se almacenan en una base de datos y se pueden ejecutar como una unidad.

Son útiles para encapsular y reutilizar lógica de negocio, simplificando el acceso y la manipulación de datos dentro de una base de datos.

<table width='99%'>
<tr>
<td bgcolor='#5b2c6f'>

## **<font color="#FFFFFF">Características</font>**

</td>
</tr>
</table>

>* **Almacenamiento en la Base de Datos**: Se guardan en el servidor de la base de datos y se pueden ejecutar cuando sea necesario.
>* **Parámetros**: Pueden aceptar parámetros de entrada y salida, lo que permite dinámicamente modificar su comportamiento.
>* **Control de Transacciones**: Pueden incluir manejo de transacciones para asegurar la integridad de los datos.
>* **Seguridad**: Permiten restringir el acceso directo a las tablas y proporcionar una interfaz controlada.
>* **Optimización y Rendimiento**: Pueden ser precompilados, lo que mejora el rendimiento en comparación con ejecutar instrucciones SQL directas.

<table width='99%'>
<tr>
<td bgcolor='#5b2c6f'>

## **<font color="#FFFFFF">Beneficios</font>**

</td>
</tr>
</table>

>* **Reutilización de Código**: Facilitan la escritura de código más limpio y mantenible.
>* **Reducción de Overhead**: Disminuyen la cantidad de datos que se envían entre el servidor de aplicaciones y la base de datos.
>* **Consistencia**: Aseguran la ejecución de la misma lógica en diferentes partes de una aplicación.
>* **Facilitación del Mantenimiento**: La lógica de negocio se puede actualizar en un solo lugar sin modificar la aplicación completa.

# **<font color="#FFFFFF">Diferencia entre Scripts, Vistas y Procedimientos Almacenados</font>**

| Característica                | Scripts                          | Vistas                                  | Procedimientos Almacenados         |
|-------------------------------|----------------------------------|-----------------------------------------|------------------------------------|
| **Definición**                | Conjuntos de instrucciones SQL.  | Consultas almacenadas en la base de datos. | Conjuntos de instrucciones SQL almacenadas. |
| **Propósito**                 | Ejecutar tareas específicas o consultas ad-hoc. | Presentar datos de manera simplificada o personalizada. | Ejecutar lógica de negocio repetitiva. |
| **Almacenamiento**            | No se almacenan (ejecución temporal). | Se almacenan en la base de datos como un objeto. | Se almacenan en la base de datos como un objeto. |
| **Parámetros**                | No aceptan parámetros.           | No aceptan parámetros.                  | Pueden aceptar parámetros de entrada y salida. |
| **Resultado**                 | Resultado inmediato de la ejecución. | Filas y columnas basadas en la consulta subyacente. | Puede devolver resultados o nada; depende de la lógica. |

| **Seguridad**                 | Limitada (el usuario tiene control total). | Proporcionan un nivel de seguridad al ocultar las tablas subyacentes. | Pueden restringir el acceso a las tablas. |
| **Uso**                       | Ideal para operaciones únicas o no repetitivas. | Útil para simplificar consultas complejas. | Ideal para encapsular lógica de negocio y mejorar la mantenibilidad del código. |

### **<font color="#FFFFFF">SET NOCOUNT ON;</font>**

</td>
</tr>
</table>

La instrucción SET NOCOUNT ON; se utiliza en SQL Server para controlar el comportamiento de las sesiones y cómo se informan los recuentos de filas afectadas por las operaciones DML (Data Manipulation Language), como INSERT, UPDATE y DELETE.

>* **Previene Mensajes de Recuento**: Cuando se ejecutan sentencias SQL, SQL Server devuelve información sobre el número de filas afectadas. Por defecto, después de cada operación, se envía un mensaje que indica cuántas filas fueron afectadas. **SET NOCOUNT ON;** suprime estos mensajes.
>* **Mejora el Rendimiento**: Al suprimir estos mensajes, se reduce la cantidad de información que SQL Server necesita enviar al cliente. Esto puede mejorar el rendimiento, especialmente en procedimientos almacenados que realizan múltiples operaciones de modificación de datos.
>* **Menos Ruido en Salida**: En procedimientos almacenados o scripts donde se realizan múltiples operaciones, puede ser útil no recibir recuentos de filas para mantener la salida más limpia y enfocada en los resultados importantes.
>* **Evitar Problemas de Interacción**: En ciertos contextos, como aquellos que utilizan cursores o en aplicaciones que procesan resultados de varias consultas, obtener un recuento de filas podría generar confusión o interferir con la lógica de control.




```
USE Northwind;
GO
/*
Obtener los pedidos por cliente
*/
CREATE OR ALTER PROCEDURE usp_pedido_x_cliente
	@CustomerID nchar(5)
AS
BEGIN
	SET NOCOUNT ON;
	SELECT
		c.CustomerID,
		c.CompanyName,
		o.OrderID,
		CONVERT(CHAR(10), o.OrderDate, 103) AS [Fecha del Pedido],
		CONVERT(CHAR(10), o.RequiredDate, 103) AS [Fecha de Requerimiento],
		CONVERT(CHAR(10), o.RequiredDate, 103) AS [Fecha de Entrega],
		o.Freight,
		o.ShipCity AS Ciudad,
		o.ShipCountry AS País
	FROM Customers c
	INNER JOIN Orders o
	ON c.CustomerID = o.CustomerID
	WHERE c.CustomerID = @CustomerID;
END;
GO
usp_pedido_x_cliente 'ANTON'
GO

```

# **<font color="#FFFFFF">Valores por defecto (DEFAULT)</font>**

</td>

Los valores por defecto permiten especificar que ciertos parámetros pueden recibir un valor predefinido si no se especifica uno al momento de invocar el procedimiento.

Esto puede simplificar la llamada a procedimientos y facilitar su uso.

<table width='99%'>
<tr>
<td bgcolor='#5b2c6f'>

## **<font color="#FFFFFF">Características de los Valores por Defecto en Parámetros</font>**

</td>
</tr>
</table>

>* **Facilitan la Llamada a Procedimientos**: Puedes llamar a un procedimiento sin proporcionar todos los parámetros, ya que aquellos que tienen valores por defecto se pueden omitir.
>* **Flexibilidad**: Permiten que una misma procedimiento se use en diferentes contextos, proporcionando valores alternativos según sea necesario.
>* **Valores de Diferentes Tipos**: Puedes establecer valores por defecto para parámetros de diversos tipos de datos (números, cadenas, fechas, etc.).



```
USE Northwind;
GO
-- Obtener la lista de nombres de productos y sus precios
-- unitarios mayor al stock mínimo de 10 unidades,
-- ordenados por stock en forma descendente.

CREATE OR ALTER PROCEDURE dbo.usp_productos_stock_minimo
(
	@StockMinimo INT = 10
)
AS
BEGIN
	SET NOCOUNT ON;
	SELECT
		ProductID,
		ProductName,
		UnitPrice,
		UnitsInStock
	FROM
		dbo.Products
	WHERE
		UnitsInStock >= @StockMinimo
	ORDER BY
		UnitsInStock DESC;
END;
GO
-- Ejecución del procedimiento almacenado para verificar su funcionamiento
EXEC dbo.usp_productos_stock_minimo @StockMinimo = 20;
GO
```

```
USE Northwind;
GO
/*
Obtener los pedidos por país con un límite de resultados.
*/
CREATE OR ALTER PROCEDURE usp_pedidos_x_pais_con_limite
	@Pais nvarchar(15),
	@Limite INT = 10 -- Valor por defecto: 10 pedidos
AS
BEGIN
	SET NOCOUNT ON;
	SELECT
		c.CompanyName,
		o.OrderID,
		CONVERT(CHAR(10), o.OrderDate, 103) AS [Fecha del Pedido],
		o.ShipCity AS Ciudad,
		o.ShipCountry AS País
	FROM Customers c
	INNER JOIN Orders o
	ON c.CustomerID = o.CustomerID
	WHERE o.ShipCountry = @Pais
	ORDER BY o.OrderID DESC
	OFFSET 0 ROWS FETCH NEXT @Limite ROWS ONLY;
END;
GO

-- Ejecución del procedimiento almacenado para verificar su funcionamiento
-- Ejemplo 1: Mostrar los 5 pedidos más recientes de 'USA'
EXEC usp_pedidos_x_pais_con_limite @Pais = 'USA', @Limite = 5;
GO

-- Ejemplo 2: Mostrar los 10 pedidos más recientes (por defecto) de 'Germany'
EXEC usp_pedidos_x_pais_con_limite @Pais = 'Germany';
GO

```

```
/* Created by GitHub Copilot in SSMS - review carefully before executing */
/*
  Procedimiento almacenado: devuelve pedidos filtrados por país de envío y limitado por número de filas.
  Parámetros:
    @ShipCountry: país de envío a filtrar. NULL = sin filtro (devuelve todos los países).
    @Top: número máximo de filas a devolver.
*/
CREATE OR ALTER PROCEDURE dbo.usp_GetOrdersByShipCountry
    @ShipCountry NVARCHAR(15) = NULL,
    @Top INT = 10 -- 10 es el valor por defecto (número máximo de filas)
AS
BEGIN
    SET NOCOUNT ON;

    -- Si se proporciona un valor inválido para @Top,
    -- restablecer al valor predeterminado
    IF @Top <= 0 SET @Top = 10; -- 10: default rows

    SELECT TOP(@Top)
        o.OrderID,
        o.CustomerID,
        c.CompanyName,
        o.OrderDate,
        o.ShippedDate,
        o.Freight,
        o.ShipCountry
    FROM dbo.Orders AS o
    LEFT JOIN dbo.Customers AS c ON o.CustomerID = c.CustomerID
    WHERE
    (
            @ShipCountry IS NULL
        OR
            o.ShipCountry = @ShipCountry
    )
    ORDER BY o.OrderDate DESC;
END;
GO
usp_GetOrdersByShipCountry 'USA', 5;
GO
```

```
USE [pvl]
GO
CREATE OR ALTER PROCEDURE usp_beneficiario_consulta
(
    @parametro VARCHAR(120) = NULL
)
AS
BEGIN
    SET NOCOUNT ON;
    IF @parametro IS NULL OR LTRIM(RTRIM(@parametro)) = ''
        SELECT
            [cod_ben_in] AS Código
            ,[pat_ben_vc] AS Paterno
            ,[mat_ben_vc] AS Materno
            ,[nom_ben_vc] AS Nombre
            ,[dni_ben_ch] AS DNI
            ,CONVERT(CHAR(10), [fec_nac_ben_da], 103) AS [Fecha de Nacimiento]
            ,IIF([cod_sex_bi] = 0, 'FEMENINO', 'MASCULINO') AS Sexo
        FROM [dbo].[Beneficiario]
        ORDER BY Paterno, Materno, Nombre
    ELSE
        IF ISNUMERIC(@parametro) = 1 AND @parametro NOT LIKE ('[0-9][0-9][0-9][0-9][0-9][0-9][0-9][0-9]')
            SELECT
                [cod_ben_in] AS Código
                ,[pat_ben_vc] AS Paterno
                ,[mat_ben_vc] AS Materno
                ,[nom_ben_vc] AS Nombre
                ,[dni_ben_ch] AS DNI
                ,CONVERT(CHAR(10), [fec_nac_ben_da], 103) AS [Fecha de Nacimiento]
                ,IIF([cod_sex_bi] = 0, 'FEMENINO', 'MASCULINO') AS Sexo
            FROM [dbo].[Beneficiario]
            WHERE [cod_ben_in] = CAST(@parametro AS INT)
            ORDER BY Paterno, Materno, Nombre
        ELSE
            IF @parametro LIKE ('[0-9][0-9][0-9][0-9][0-9][0-9][0-9][0-9]')
                SELECT
                    [cod_ben_in] AS Código
                    ,[pat_ben_vc] AS Paterno
                    ,[mat_ben_vc] AS Materno
                    ,[nom_ben_vc] AS Nombre
                    ,[dni_ben_ch] AS DNI
                    ,CONVERT(CHAR(10), [fec_nac_ben_da], 103) AS [Fecha de Nacimiento]
                    ,IIF([cod_sex_bi] = 0, 'FEMENINO', 'MASCULINO') AS Sexo
                FROM [dbo].[Beneficiario]
                WHERE [dni_ben_ch] = @parametro
                ORDER BY Paterno, Materno, Nombre
            ELSE
                IF @parametro LIKE ('[0-9][0-9]/[0-9][0-9]/[0-9][0-9][0-9][0-9]')
                    SELECT
                        [cod_ben_in] AS Código
                        ,[pat_ben_vc] AS Paterno
                        ,[mat_ben_vc] AS Materno
                        ,[nom_ben_vc] AS Nombre
                        ,[dni_ben_ch] AS DNI
                        ,CONVERT(CHAR(10), [fec_nac_ben_da], 103) AS [Fecha de Nacimiento]
                        ,IIF([cod_sex_bi] = 0, 'FEMENINO', 'MASCULINO') AS Sexo
                    FROM [dbo].[Beneficiario]
                    WHERE CONVERT(CHAR(10), [fec_nac_ben_da], 103) = @parametro
                    ORDER BY Paterno, Materno, Nombre
                ELSE
                    SELECT
                        [cod_ben_in] AS Código
                        ,[pat_ben_vc] AS Paterno
                        ,[mat_ben_vc] AS Materno
                        ,[nom_ben_vc] AS Nombre
                        ,[dni_ben_ch] AS DNI
                        ,CONVERT(CHAR(10), [fec_nac_ben_da], 103) AS [Fecha de Nacimiento]
                        ,IIF([cod_sex_bi] = 0, 'FEMENINO', 'MASCULINO') AS Sexo
                    FROM [dbo].[Beneficiario]
                    WHERE CONCAT([pat_ben_vc], SPACE(1), [mat_ben_vc], SPACE(1), [nom_ben_vc]) LIKE @parametro + '%'
                    ORDER BY Paterno, Materno, Nombre
  RETURN 0
END
GO
usp_beneficiario_consulta 'c'
GO
```

```
Imports System.Data.SqlClient
Public Class FrmPrincipal
    Private ObjSqlConnection As New SqlConnection

    Private Sub TxtBuscar_TextChanged(sender As Object, e As EventArgs) Handles TxtBuscar.TextChanged
        Buscar()
    End Sub

    Private Sub FrmPrincipal_Load(sender As Object, e As EventArgs) Handles MyBase.Load
        Buscar()
    End Sub

    Private Sub Buscar()
        Dim ObjSqlConnection As New SqlConnection
        With ObjSqlConnection
            .ConnectionString = "Data Source=server;Integrated Security=SSPI;Initial Catalog=pvl"
            .Open()
        End With
        Dim ObjSqlParameter As New SqlParameter("@parametro", Me.TxtBuscar.Text())
        Dim ObjSqlCommand As New SqlCommand()
        With ObjSqlCommand
            .Connection = ObjSqlConnection
            .CommandText = "usp_beneficiario_consulta"
            .CommandType = CommandType.StoredProcedure
            .Parameters.Add(ObjSqlParameter)
        End With
        Dim ObjSqlDataAdapter As New SqlDataAdapter(ObjSqlCommand)
        Dim objDataTable As New DataTable

        ObjSqlDataAdapter.Fill(objDataTable)

        With DgvBeneficiario
            .DataSource = objDataTable
            .AllowDrop = False
            .AllowUserToAddRows = False
            .AllowUserToDeleteRows = False
            .AllowUserToOrderColumns = False
            .AllowUserToResizeColumns = False
            .AllowUserToResizeRows = False
            .Columns(0).Width = 50      ' Código
            .Columns(1).Width = 150     ' Paterno
            .Columns(2).Width = 150     ' Materno
            .Columns(3).Width = 200     ' Nombres
            .Columns(4).Width = 65      ' DNI
            .Columns(5).Width = 75      ' Fecha de Nacimiento
            .Columns(6).Width = 85      ' Sexo
            .EditMode = DataGridViewEditMode.EditProgrammatically
        End With
        ObjSqlConnection.Close()
    End Sub
End Class
```
